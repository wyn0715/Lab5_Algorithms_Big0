```python
LAST_NAME = "Wales"
STUDENT_ID = "TUPM-26-7925"
FAVORITE_ARTIST = "IVOS"

id_digits = [int(c) for c in STUDENT_ID if c.isdigit()]
letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]

SEED_DIGIT = id_digits[-1]
ID_SUM = sum(id_digits)
NAME_LENGTH = len([c for c in LAST_NAME if c.isalpha()])
ARTIST_LENGTH = len(letters)
ARTIST_SUM = sum(ord(c) - ord("A") + 1 for c in letters)

print("SEED_DIGIT :", SEED_DIGIT)
print("ID_SUM :", ID_SUM)
print("NAME_LENGTH :", NAME_LENGTH)
print("ARTIST_LENGTH :", ARTIST_LENGTH)
print("ARTIST_SUM :", ARTIST_SUM)
```

```python
import random

n = 200 + ID_SUM
seed_value = int("".join(str(d) for d in id_digits)) + ARTIST_SUM
rng = random.Random(seed_value)

base = rng.sample(range(1000, 1000 + n * 10), n)     

conditions = {
    "Random":  base,
    "Sorted":  sorted(base),
    "Reverse": sorted(base, reverse=True),
}
expected = sorted(base)

print("Number of orders:", n)
print("First 10 random orders :", conditions["Random"][:10])
print("First 10 sorted orders :", conditions["Sorted"][:10])
print("First 10 reverse orders:", conditions["Reverse"][:10])
```

```python
def bubble_sort(data):
    a, comps, moves = data[:], 0, 0
    n = len(a)
    for i in range(n - 1):
        swapped = False
        for j in range(n - 1 - i):
            comps += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                moves += 1
                swapped = True
        if not swapped:          
            break
    return a, comps, moves
```

```python
def selection_sort(data):
    a, comps, moves = data[:], 0, 0
    n = len(a)
    for i in range(n - 1):
        m = i
        for j in range(i + 1, n):
            comps += 1
            if a[j] < a[m]:
                m = j
        if m != i:
            a[i], a[m] = a[m], a[i]
            moves += 1
    return a, comps, moves
```

```python
def insertion_sort(data):
    a, comps, moves = data[:], 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comps += 1
            if a[j] > key:
                a[j + 1] = a[j]
                moves += 1
                j -= 1
            else:
                break
        a[j + 1] = key
    return a, comps, moves
```

```python
import time

algorithms = (("Bubble Sort", bubble_sort),
              ("Selection Sort", selection_sort),
              ("Insertion Sort", insertion_sort))

results = []
for cond, data in conditions.items():
    for name, func in algorithms:
        t0 = time.perf_counter()
        result, comps, moves = func(data)
        elapsed = time.perf_counter() - t0
        results.append((cond, name, comps, moves, elapsed, result == expected))

print(f"{'Condition':<10}{'Algorithm':<16}{'Comparisons':>12}{'Movements':>11}{'Time (s)':>12}  Verified")
print("-" * 70)
for i, (cond, name, comps, moves, elapsed, ok) in enumerate(results):
    print(f"{cond:<10}{name:<16}{comps:>12}{moves:>11}{elapsed:>12.6f}  {'PASS' if ok else 'FAIL'}")
    if (i + 1) % 3 == 0:
        print()
```

```python
all_ok = all(r[5] for r in results)

print("Sorting verification:", "ALL PASSED" if all_ok else "FAILED")
print()
print("Big O analysis:")
print("  Bubble    best O(n) [early exit] | average O(n^2) | worst O(n^2)")
print("  Selection best O(n^2)            | average O(n^2) | worst O(n^2)")
print("  Insertion best O(n)              | average O(n^2) | worst O(n^2)")
print()
print("Best-case  : Sorted input  (Bubble and Insertion need about n comparisons)")
print("Worst-case : Reverse input (all three need about n^2/2 comparisons)")
print("Average    : Random input")
print()
print("Final Output: all three algorithms sorted every input condition correctly.")
```
